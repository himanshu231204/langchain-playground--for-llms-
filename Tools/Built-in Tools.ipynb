Built-in tool - Shell tool


The Shell Tool in LangChain allows an LLM (via an Agent) to execute system shell commands on your machine.

It gives the LLM the ability to interact with the operating system.

In simple words:

Shell Tool = A tool that allows the LLM to run terminal/command-line commands.

```python
from langchain_community.tools import ShellTool

shell_tool = ShellTool()

result = shell_tool.invoke("ls")

print(result)
```
